```python
# Time Complexity O(n):
def linear_search(arr, search_element):
    for index, element in enumerate(arr):
        if element == search_element:
            return index
    return None
```

```python
# Time Complexity O(log n): 
# Works only on sorted array
def binary_search(arr, search_elem):
    start = 0
    end = len(arr) - 1

    while start <= end:
        mid = (start + end) // 2
        if arr[mid] == search_elem:
            return mid  # Element found, return index
        elif arr[mid] < search_elem:
            start = mid + 1  # Search right half
        else:
            end = mid - 1  # Search left half

    return -1  # Element not found
```

```python
def bubble_sort(arr):
    for i in range(0, len(arr)):
        for j in range(i, len(arr)):
            if arr[i] > arr[j]:
                arr[j], arr[i] = arr[i], arr[j]
    return arr
bubble_sort([3,2,7,4,9,6,8,9,2,1])
```

```python
def bubble_sort_with_temp(arr):
    temp = None
    for i in range(0, len(arr)):
        for j in range(i, len(arr)):
            if arr[i] > arr[j]:
                temp = arr[i]   # Above example using `temp` variable
                arr[i] = arr[j]
                arr[j] = temp
    return arr
bubble_sort([3,2,7,4,9,6,8,9,2,1])
```

```python
def merge(left, right):
    result = []
    i = j = 0
    
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
            
    result.extend(left[i:])
    result.extend(right[j:])
    
    return result

def mergeSort(arr):
    step = 1  # Starting with sub-arrays of length 1
    length = len(arr)
    
    while step < length:
        for i in range(0, length, 2 * step):
            left = arr[i:i + step]
            right = arr[i + step:i + 2 * step]
            
            merged = merge(left, right)
            
            # Place the merged array back into the original array
            for j, val in enumerate(merged):
                arr[i + j] = val
                
        step *= 2  # Double the sub-array length for the next iteration
        
    return arr

unsortedArr = [3, 7, 6, -10, 15, 23.5, 55, -13]
sortedArr = mergeSort(unsortedArr)
print("Sorted array:", sortedArr)
```
