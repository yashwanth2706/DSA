```python
def move_zeros(arr):
    update_index = 0
    for i in range(len(arr)):
        if arr[i] != 0:                   # If the number is non-zero
            arr[update_index] = arr[i]    # update it to current update_index position remove all zeros currently since it's non changing we can update later
            update_index += 1             # once the updated, increment the update_index to add the non-zero element in the next index
            
    while update_index < len(arr):        # check if update_index matches the length of arr
        arr[update_index] = 0             # assign zero to all the remaining positions of the arr
        update_index += 1                 # increment the update_index
    return arr
```

```python
move_zeros([0,0,0,0,0,0,0,-1,-3,0,4,6,7,9])
```

```python
def move_ones(arr):
    update_index = 0
    for i in range(len(arr)):
        if arr[i] != 1:
            arr[update_index] = arr[i]
            update_index += 1
    
    while update_index < len(arr):
        arr[update_index] = 1
        update_index += 1
        
    return arr
    
```

```python
move_ones([1,1,1,1,1,1,1,-1,-3,1,4,6,7,9])
```

```python
# Boyer-Moore Voting Algorithm for majority element of O(1)
def majority_element(nums):
    vote = 0
    candidate = nums[vote] # Assume First element in the array or it can be None
    
    for new_candidate in nums:
        if vote == 0:               # check if there is no vote abandond current candidate to pick newone
            vote = 1                # give minimum vote for new candidate to participate
            candidate = new_candidate # pick new candidate
        elif new_candidate == candidate: # if new candidate is same as current candidate 
            vote += 1                    # increase the vote
        else:                            # if new candidate is not same as current candidate
            vote -= 1                    # decrease the vote 
            
    return candidate
```

```python
majority_element([1,2,3,4,1,1,1,0,3,4,1,3])
```
