Algorithm to remove duplicate from the array
* Array Type: Sorted
* In a sorted array we can remove the duplicates without using extra memory

```python
def remove_duplicate_without_hash(nums):# For sorted array the optimal approach O(n) without O(n^2) time
    length = len(nums) + 1              # Memory O(1)
    pointer = 0
    for index in range(1, length):
        if nums[index-1] != nums[pointer]:
            pointer += 1
            nums[pointer] = nums[index-1]
    nums = nums[:pointer+1] # we can use this method to replace the current array
    # pointer + 1 becasue it is starting with zero and list slicing [:7] up to 6 only
    # OR
    # while pointer < length-2: # lenght is +1 pointer is starting from 0 so decrease by 2
    #    nums.pop()
    #    pointer += 1
    return nums  
```

Algorithm to remove duplicate from the array
* Array Type: Un-Sorted
* Hashmap needs to be implemented if the array is unsorted

```python
def remove_duplicate_with_hash(nums): # For unsorted array O(n) but with extra memory O(n)
    seen = set()
    for num in nums:
        if not num in seen:
            seen.add(num)
    nums = list(seen)
    return nums
```

```python
remove_duplicate_with_hash([1,1,1,1,1,1,2,2,2,3,3,4,4,4,5,5,5,5,5,6])
```
