String Slicing

```python
alphabets = "abcdefghijk"
alphabets
```

```python
alphabets[2]
```

```python
alphabets[2:] # Starts from 2-index and goes all the way to the end of the string
```

```python
alphabets[:3] # Starts from beggining (0-index) upto 3-index (does not include 3rd index)
```

Syntax: 
    string[start:stop]

```python
alphabets[3:6] # start from 3 upto 6 to get 'def' from 'abcdefghijk'
```

Step Size

string[start:stop:stepsize]

```python
alphabets[::] # All the way from beggining to all the way to the end
```

```python
alphabets[::2] # go from beggining to end in jumps of 2
```

```python
alphabets[::3] # go from beggining to end in jumps of 3
```

```python
alphabets[2:7:2] # start from 2nd-index upto 7th (does not include 7th) jumps of 2
# Jumps of 2 skips a single charecter
```

```python
alphabets[::-1] # reverse indexing with reverse string
```
