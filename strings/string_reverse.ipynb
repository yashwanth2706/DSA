```python
def reverse(string):
    length = len(string)
    reverse_string = ""
    for i in range(1, length+1):
        reverse_string += string[length-i]
    return reverse_string
```

```python
reverse("yashwanth")
```

```python
def reverse_with_negative_index(string):
    length = len(string) + 1
    reverse = ""
    for i in range(1, length):
        reverse += string[-i]
    return reverse
```

```python
reverse_with_negative_index("yashwanth")
```

```python
def reverse_string_with_string_slicing(string):
    return string[::-1]
```

```python
reverse_string_with_string_slicing("yashwanth")
```
