```python
def search_and_return_first_indexof_substring(substring, string):
    substring_length = len(substring)
    for i in range(len(string)):
        if string[i:i+substring_length] == substring:
            return i
    return -1
```

```python
search_and_return_first_indexof_substring("etc", "leetcode")
```

```python
# without string slicing
def search_substring_without_string_slicing(substring, string):
    def is_substring(current_index):
        sub_string_index = 0
        while sub_string_index < len(substring):  
            if string[current_index] != substring[sub_string_index]:
                return False
            sub_string_index += 1
            current_index += 1
        return True
    for i in range(len(string)):
        if is_substring(i):
            return i
    return -1
```

```python
search_substring_without_string_slicing("code", "coccocodcode")
```
