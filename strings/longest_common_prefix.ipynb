* Input strs = ["flower","flow","flight"]
* Output: "fl"
* If not return ""

```python
def longest_common_prefix(strs):
    minimum_length = len(min(string for string in strs)) # Get minimum length from given strs list
    prefix = ""
    for i in range(minimum_length):
        current_char = strs[0][i]
        for string in strs:
            if current_char != string[i]:
                return prefix
        prefix += string[i]
    return prefix
```

```python
longest_common_prefix(["google","goog","googly","googles"])
```
