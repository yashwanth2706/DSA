```python
def minimum(braces):
    opening = '{'
    closing = '}'
    open = 0
    close = 0
    
    if len(braces) == 1:
        return -1
    
    for char in braces:
        if char == opening:
            open += 1
        elif char == closing:
            close += 1
            
    minimum = open - close
    
    if minimum < 0:
        return -minimum
    else:
        return minimum
```

```python
minimum('{}{{}}}}')
```

```python
def square(nums):
    squared = [n**2 for n in nums]
    squared.sort()
    for num in squared:
        print(num, end=' ')
square([-2, 0, -1, 1, 2, 4, 3])
```

```python
def square(nums):
    squared = list(map(lambda x: x**2, nums))
    squared.sort()
    for num in squared:
        print(num, end=' ')
square([-2, 0, -1, 1, 2, 4, 3])
```

```python
class Solution:
    def change(self, amount: int, coins: list) -> int:
        # We can solve this problem using dynamic programming
        # We initialize array with the amounts range
        dp = [0] * (amount + 1)
        # We keep 0th index as base case
        dp[0] = 1

        # We start from each coin and start counting the permutations
        for coin in coins:
            for count in range(coin, amount + 1):
                dp[count] += dp[count - coin]
        return dp[amount]
```

```python
sol = Solution()
sol.change(5, [1,2,4])
```
