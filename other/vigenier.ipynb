```python
def encrypt(message, key):
    matrix = "abcdefghijklmnopqrstuvwxyz"

    # Normalize: lowercase both and strip spaces
    message = message.lower()
    key = "".join([c.lower() for c in key if c.isalpha()])

    # Extend key to match message length
    normalized_key = (key * (len(message) // len(key) + 1))[:len(message)]

    encrypted_msg = ""

    for i, ch in enumerate(message):
        if ch not in matrix:
            encrypted_msg += ch
            continue

        # Encrypt using Vigenère logic
        row = matrix.index(ch)
        col = matrix.index(normalized_key[i])
        encrypted_char = matrix[(row + col) % 26]

        encrypted_msg += encrypted_char

    return encrypted_msg
            
```

```python
encrypt('attack at dawn', 'secret')
```

```python
def decrypt(message, key):
    matrix = "abcdefghijklmnopqrstuvwxyz"

    # Normalize: lowercase and strip to only alphabetic chars
    key = "".join([c.lower() for c in key if c.isalpha()])
    message = message.lower()

    # Extend key to match message length
    normalized_key = (key * (len(message) // len(key) + 1))[:len(message)]

    decrypted_msg = ""

    for i, ch in enumerate(message):
        if ch not in matrix:
            decrypted_msg += ch
            continue

        # Reverse the shift used in encryption
        row = matrix.index(ch)
        col = matrix.index(normalized_key[i])
        size = row - col
        if size < 0:
            size = size + 26
        decrypted_char = matrix[size % 26]

        decrypted_msg += decrypted_char

    return decrypted_msg
```

```python
decrypt('sxvrgd ev htor', 'secret')
```
